ADJUSTED CLOSE PRICE

```python
df = df[['Close']]
df.head()
```

VARIABLE FOR PREDICTION FUTURE PRICE

```python
forecast_out = 30
df['Prediction'] = df[['Close']].shift(-forecast_out)
```

#THE INDEPENDENT DATASET
CONVERT THE DATAFRAME TO A NUMPY ARRAY

```python
X = np.array(df.drop(['Prediction'], axis=1))
X = X[:-forecast_out]
```

#THE DEPENDENT DATASET
CONVERT THE DATAFRAME TO A NUMPY ARRAY

```python
y = np.array(df['Prediction'])
y = y[:-forecast_out]
```

SPLIT THE DATASET

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

TRAIN THE SUPPORT VECTOR MACHINE (REGRESSOR)

```python
svr_rbf = SVR(kernel='rbf', C=1e3, gamma=0.1)
svr_rbf.fit(x_train, y_train)
```

TEST THE MODEL

```python
svm_confidence = svr_rbf.score(x_test, y_test)
print("svm confidence: ", svm_confidence)
```

TRAIN THE LINEAR REGRESSION

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
```

TEST THE MODEL

```python
lr_confidence = lr.score(x_test, y_test)
print("lr confidence: ", lr_confidence)
```

SET x_FORECAST EQUAL TO THE LAST 30RAWS OF THE ORIGINAL DATASET

```python
x_forecast = np.array(df.drop(['Prediction'], axis=1))[-forecast_out:]
```

THE LINEAR REGRESSION PREDICTIONS MODEL FOR THE NEXT 'n' DAYS

```python
lr_prediction = lr.predict(x_forecast)
print(lr_prediction)
```
